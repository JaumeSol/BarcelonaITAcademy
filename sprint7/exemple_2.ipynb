# Exemple 2: Càlcul de l'IMC

🧠 **Algoritme principal**

✔ `main_imc()`:

1. Demanar dos números a l'usuari via input: `obtenir_dades()`
2. Reemplazamos , por . y convertimos les dades a números flotantes: `convertir_dades()`
3. Validamos rangos: `validar_dades()`
  - si alguno es < 0 entonces no es válido
  - si peso > 300 entonces no es válido
  - si altura > 2.5 entonces no es válido
4. Calcular l'IMC: `calcular_imc()`
5. Classificar l'IMC: `classificar_imc()`
6. Mostrar el resultat: `mostrar_resultat()`

```python
def obtenir_dades():
    pes = input("Introdueix el pes en kg: ")
    alçada = input("Introdueix la alçada en metres: ")
    return pes, alçada
```

✔ `convertir_dades(pes, alçada)`

1. Trata de:
    - Reemplazar en pes la , por .
    - Convertir el pes a float
    - Reemplazar en alçada la , por .
    - Convertir la alçada a float
    - Retornalos
2. Si falla algún punto anterior entonces mostra un missatge d'error

```python
def convertir_dades(pes, alçada):
    try:
        pes = float(pes.replace(',', '.'))
        alçada = float(alçada.replace(',', '.'))
        return pes, alçada
    except ValueError:
        print("Error: Si us plau, introdueix valors numèrics vàlids per al pes i l'alçada.")
        return None, None
```

✔ `validar_dades(pes, alçada)`

1. Si 0 < pes < 300 y 0 < alçada < 2.5 entonces retorna none
2. Si no retorna none y mostra un missatge d'error

```python
def validar_dades(pes, alçada):
    if 0 < pes < 300 and 0 < alçada < 2.5:
        return True
    else:
        print("Error: Si us plau, introdueix números vàlids.")
        return False
```

✔ `calcular_imc(pes, alçada)`

1. Calcula fórmula pes / alcada**2 y almacena en resultat.
2. Retorna resultat

```python
def calcular_imc(pes, alçada):
    return pes / (alçada ** 2)
```

✔ ``classificar_imc(imc)``
1. si imc < 18.5 → Baix pes
2. si < 25 → Normal
3. si < 30 → Sobrepès
4. si no → Obesitat

```python
def classificar_imc(imc):
    if imc < 18.5:
        return "Baix pes"
    elif imc < 24.9:
        return "Pes normal"
    elif imc < 29.9:
        return "Sobrepès"
    return "Obesitat"
```

✔ `mostrar_resultat(pes, alçada, imc, classificacio)`
1. imprimir els valors i el resultat final


```python
def mostrar_resultats(pes, alçada, imc, classificacio):
    print(f"Pes en kg: {pes}\nAlçada en m: {alçada}")
    print(f"L'IMC és {round(imc,2)} i classificat com: {classificacio}")
```

✔ `main_imc()`:

1. `obtenir_dades()` almacena en pes y alçada
2. `convertir_dades(pes, alçada)` almacena en pes y alçada
3. Si pes y alçada son números entonces:
    - Si `validar_dades(pes, alçada)` almacena un bool es true entonces:
        - `calcular_imc(pes, alçada)` almacena en imc
        - `classificar_imc(imc)` almacena en classificacio
        - `mostrar_resultats(pes, alçada, imc, classificacio)`

```python
def main_imc():
    pes, alçada = obtenir_dades()
    pes, alçada = convertir_dades(pes, alçada)

    if pes and alçada:
        if validar_dades(pes, alçada):
            imc = calcular_imc(pes, alçada)
            classificacio = classificar_imc(imc)
            mostrar_resultats(pes, alçada, imc, classificacio)
```
