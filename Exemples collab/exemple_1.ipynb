## Resultat

🧠 **Algoritme inicial (versió molt bàsica)**

1. Demanar dos números a l'usuari.
2. Sumar-los.
3. Mostrar el resultat.

**Primer desgranaje**

1. Demanar dos números a l'usuari via input.
2. Convertir aquestes dades a números flotantes.
3. Sumar-los.
4. Mostrar el resultat.

**Segundo desgranaje**

1. Demanar dos números a l'usuari via input.
2. Validar la conversion de lo dades a números flotantes.
3. Sumar-los.
4. Mostrar el resultat.


🧠 **Algoritme principal**

✔ `main_suma()`:

1. Demanar dos números a l'usuari via input: `obtenir_dades()`
2. Validar la conversion de lo dades a números flotantes: `validar_dades()`
3. Sumar-los: `sumar()`
4. Mostrar el resultat: `mostrar_resultat()`

🧩 **Desgranant el problema: naixement de les funcions**

✔ `obtenir_dades()`:

1. Demanar primer número y almacenar en numero1.
2. Demanar segundo número y almacenar en numero2.
3. Retornar numero1 y numero2.

```python
def obtenir_dades():
    numero1 = input("Introdueix el primer número: ")
    numero2 = input("Introdueix el segon número: ")
    return numero1, numero2
```

✔ `validar_dades(numero1, numero2)`

1. Trata de:
    - Convertir el numero1 a float
    - Convertir el numero2 a float
    - Retornalos
2. Si falla algún punto anterior entonces mostra un missatge d'error

```python
def validar_dades(numero1, numero2):
    try:
        num1 = float(numero1)
        num2 = float(numero2)
        return num1, num2
    except ValueError:
        print("Error: Si us plau, introdueix números vàlids.")
        return None, None
```

✔ `sumar(num1, num2)`

1. Calcula num1 + num2 y almacena en resultat.
2. Retorna resultat

```python
def sumar(num1, num2):
    resultat = num1 + num2
    return resultat
```

✔ `mostrar_resultat(num1, num2, resultat)`

1. Imprimir en pantalla "Has introduït X i Y, la seva suma és Z"


```python
def mostrar_resultat(num1, num2, resultat):
    print(f"Has introduït {num1} i {num2}, la seva suma és {resultat}")
```

🧠 **Algoritme principal**

✔ `main_suma()`:

1. `obtenir_dades()` almacena en numero1 y numero2
2. `validar_dades(numero1, numero2)` almacena en num1 y num2
3. Si num1 y num2 son números entonces:
    - `sumar(num1, num2)` almacena en resultat
    - `mostrar_resultat(num1, num2, resultat)`

```python
def main_suma():
    numero1, numero2 = obtenir_dades()
    num1, num2 = validar_dades(numero1, numero2)
    if num1 and num2:
        resultat = sumar(num1, num2)
        mostrar_resultat(num1, num2, resultat)
```

Probamos con caso de exito (10, 20) y no válido (1, ljk)
