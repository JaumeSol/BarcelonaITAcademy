La funció obtenir_dades ha de recollir la següent informació:
Nom i Cognoms
Telèfon de contacte
Anys d'experiència dins de l'empresa
Coneixement de Català (0–3 punts)

    0 = sense coneixement
    1 = fins a nivell B1
    2 = nivell B2
    3 = C1 o superior

Domini d'Excel (0–3 punts)

    0 = nivell bàsic
    1 = ús de funcions
    2 = taules dinàmiques
    3 = Visual Basic i Macros

Títol acadèmic (0–3 punts)

    0 = ESO
    1 = FP
    2 = Universitat
    3 = Màster o superior



```python
def obtenir_dades():
  """recollir les dades de l'usuari."""
  nom_cog = input("Nom i cognom? ")
  tel = input("Telefon contacte? ")
  anys = int(input("Anys experiencia? "))
  print("Nivell català?\n> ")
  nivell_cat = int(input("0 quan és nul, 1 fins B1, 2 fins B2, 3 per C1 o més - "))
  print("Nivell d'Excel?\n> ")
  nivell_excel = int(input("0 per bàsic, 1 funcions, 2 taules dinàmiques, 3 VB i Macros - "))
  print("Formació? ")
  forma = int(input("0 per ESO, 1 per FP, 2 per Universitat, 3 per Master o més - "))
  return [nom_cog, tel, anys, nivell_cat, nivell_excel, forma]
```

Ara hem de fer la funció de validació de les dades:


```python
def validar_dades(nom_cog, tel, anys, nivell_cat, nivell_excel, forma):
  """la funció espera 6 paràmetres que revisem"""
  if not str(nom_cog).strip():
    print("Nom incorrecte ")
    return False
  if not len(tel).strip() == 9:
    print("Numero incorrecte ")
    return False
  if not 0 <= anys <= 52:
    print("anys experiencia incorrectes")
    return False
  if not 0 <= nivell_cat <= 3:
    print("nivell català fora de rang")
    return False
  if not 0 <= nivell_excel <= 3:
    print("nivell excel fora de rang")
    return False
  if not 0 <= forma <= 3:
    print("nivell formació icorrecte")
    return False
  else:
    return True
```
